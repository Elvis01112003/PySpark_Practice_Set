```python
# Create emp DataFrame

emp = spark.createDataFrame(data=emp_data, schema=emp_schema)
```

```python
# Get number of partitions and show data

from pyspark.sql.functions import spark_partition_id

# Use Spark SQL to get partition count (Spark Connect compatible)
num_partitions = emp.select(spark_partition_id()).distinct().count()
print(f"Number of partitions: {num_partitions}")
emp.show()
```

```python
# Write the data in parquet format

emp.write.format("parquet").mode("overwrite").save("/Volumes/workspace/default/orders/emp.parquet")
```

```python
# View data partition information
from pyspark.sql.functions import spark_partition_id

emp.withColumn("partition_id", spark_partition_id()).show()
```

```python
emp.write.format("csv").option("header", True).mode("overwrite").save("/Volumes/workspace/default/orders/emp.csv")
```

```python
# Write the data with Partition to output location

emp.write.format("csv").partitionBy("department_id").option("header", True).mode("overwrite").save("/Volumes/workspace/default/orders/emp_partitioned")
```
