```python
# Reading with Schema

_schema = "dept_id int, dept_name string, location string"

df = spark.read.format("csv").option("header", "true").schema(_schema).load("/Workspace/Users/miketyson.rocks69@gmail.com/PySpark_Practice_Set/practice/departments.csv")
```

```python
# Handle BAD records - PERMISSIVE (Default mode)

_schema = "dept_id int, dept_name string, location string, bad_record string"

df_p = spark.read.format("csv").schema(_schema).option("columnNameOfCorruptRecord", "bad_record").option("header", True).load("/Workspace/Users/miketyson.rocks69@gmail.com/PySpark_Practice_Set/practice/departments.csv")
```

```python
df_p.printSchema()
```

```python
df_p.show()
```

```python
# Handle BAD records - DROPMALFORMED
_schema = "dept_id int, dept_name string, location string"

df_m = spark.read.format("csv").option("header", True).option("mode", "DROPMALFORMED").schema(_schema).load("/Workspace/Users/miketyson.rocks69@gmail.com/PySpark_Practice_Set/practice/departments.csv")
```

```python
df_m.printSchema()
```

```python
df_m.show()
```

```python
# Handle BAD records - FAILFAST
# Note: FAILFAST will fail if there are any NULL values or malformed records
# This CSV has NULL values, so FAILFAST will throw an error when we try to show() the data
# Keeping FAILFAST for demonstration purposes - the error will occur on .show()

_schema = "dept_id int, dept_name string, location string"

df_f = spark.read.format("csv").option("header", True).option("mode", "FAILFAST").schema(_schema).load("/Workspace/Users/miketyson.rocks69@gmail.com/PySpark_Practice_Set/practice/departments.csv")
```

```python
df_f.printSchema()
```

```python
try:
    df_f.show()
except Exception as e:
    print(f"Error: {str(e).split(':')[0]}")
    print("FAILFAST mode fails when CSV has NULL/empty values")
```
