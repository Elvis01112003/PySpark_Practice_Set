```python
%sql
CREATE DATABASE sales_db;
CREATE DATABASE product_db;
```

```python
%sql
CREATE TABLE sales_db.customers (
    customer_id INT,
    customer_name STRING,
    city STRING
);
```

```python
%sql
CREATE TABLE sales_db.orders (
    order_id INT,
    customer_id INT,
    product_id INT,
    amount DECIMAL(10,2)
);
```

```python
%sql
CREATE TABLE product_db.products (
    product_id INT,
    product_name STRING,
    category STRING
);
```

```python
%sql
INSERT INTO product_db.products VALUES
(201, 'Laptop', 'Electronics'),
(202, 'Mouse', 'Electronics'),
(203, 'Chair', 'Furniture'),
(204, 'Keyboard', 'Electronics'),
(205, 'Desk Lamp', 'Home');
```

```python
%sql
SELECT 
    c.customer_name,
    o.order_id,
    p.product_name,
    p.category,
    o.amount
FROM sales_db.customers c
JOIN sales_db.orders o
    ON c.customer_id = o.customer_id
JOIN product_db.products p
    ON o.product_id = p.product_id;
```
