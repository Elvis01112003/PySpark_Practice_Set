```python
# Read the existing products table from product_db
products_df = spark.table("product_db.products")

# Copy the products table into sales_db
# overwrite mode replaces the table if it already exists
products_df.write.mode("overwrite").saveAsTable("sales_db.products")
```

```python
# Verify the products table in sales_db
spark.table("sales_db.products").show()
```
