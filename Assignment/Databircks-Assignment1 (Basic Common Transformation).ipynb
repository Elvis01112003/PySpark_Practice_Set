<b>How many employees are in the input file?

```python
# Write your code using the Spark Function
df = spark.read.csv("/Workspace/Users/miketyson.rocks69@gmail.com/PySpark_Practice_Set/Assignment/EmployeeData.csv", header=True, inferSchema=True)
cnt=df.count()
print("Number of employees in the dataset: ",cnt)
```

<b> What is the average salary of all the employees?

```python
# Write your code using the Spark Function
from pyspark.sql.functions import avg, round
df.select(round(avg("Salary"), 2).alias("Average_Salary")).show()
```

<b> How many male and female employees are there in the input file?

```python
# Write your code using the Spark Function
# df.groupBy('Gender').count().alias('Gender_Cnt').orderBy('Gender').show()
from pyspark.sql import functions as F

df.groupBy("Gender").agg(F.count("*").alias("Employee_Count")).orderBy('gender').show()
```

<b> Who is the employee with the highest salary?

```python
# Write your code using the Spark Function
from pyspark.sql import functions as f
df.orderBy(f.col('salary').desc()).show(1)
```

<b> What is the total bonus amount paid to all the employees?

```python
# Write your code using the Spark Function
from pyspark.sql import functions as F

df.select(F.sum("Bonus").alias("Total_Bonus")).show()
```

<b> How many employees were hired in the year 2020?

```python
# Write your code using the Spark Function
from pyspark.sql import functions as F

hire=df.filter(F.year("creationDate") == 2020).count()
print("Number of employees hired in 2020: ", hire)
```

<b> What is the total salary paid to all male employees?

```python
# Write your code using the Spark Function
from pyspark.sql import functions as F

df.filter(F.col("Gender") == "Male").select(F.sum("Salary").alias("Total_Male_Salary")).show()
```

```python
# Write your code using the Spark Function# Write your code using the Spark Function
df.sort("bonus", ascending=False).select(
    "id", "email", "bonus").show(1)
```

<b> What is the earliest creation date of any employee in the input file?

```python
# Write your code using the Spark Function
from pyspark.sql import functions as F

df.select(
    F.min("creationDate").alias("Earliest_Creation_Date")
).show()
```

<b> How many employees have a salary greater than 100,000?

```python
# Write your code using the Spark Function
sl=df.filter(
    df["salary"] > 100000
).count()
print("Number of employees with salary greater than 100000: ", sl)
```

<b> Who is the female employee with the highest salary?

```python
# Write your code using the Spark Function
df.filter(
    df["gender"] == "Female"
).sort(
    "salary", ascending=False
).show(1)
```

<b> What is the total bonus amount paid to female employees?

```python
# Write your code using the Spark Function
from pyspark.sql import functions as F

df.filter(
    F.col("gender") == "Female"
).select(
    F.sum("bonus").cast("decimal(10,2)").alias("Total_Female_Bonus")
).show()
```

<b> How many employees have a salary between 50,000 and 75,000?

```python
# Write your code using the Spark Function
df.filter(
    (F.col("salary") >= 50000) &
    (F.col("salary") <= 75000)).count()
    # .select("id","salary").show()
```
