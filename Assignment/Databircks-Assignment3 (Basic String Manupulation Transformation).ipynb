<b> How many employees have a first name that starts with the letter 'J'?

```python
# Write your code using the Spark Function
from pyspark.sql.functions import col

df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("/Workspace/Users/miketyson.rocks69@gmail.com/PySpark_Practice_Set/Assignment/EmployeeData.csv")

result = df.filter(col("first_name").startswith("J")).count()

print(result)
```

<b> How many employees have a last name that ends with the letter 's'?

```python
# Write your code using the Spark Function
df.filter(col("last_name").endswith("s")).count()
```

<b> What is the length of the longest first name?

```python
# Write your code using the Spark Function
from pyspark.sql.functions import length, max

df.select(max(length("first_name")).alias("longest_name_length")).show()
```

d
<b> What is the length of the shortest last name?

```python
# Write your code using the Spark Function
from pyspark.sql.functions import length, min

df.select(min(length("last_name")).alias("shortest_last_name_length")).show()
```

<b> What is the email domain of the employee with the highest salary?

```python
# Write your code using the Spark Function
from pyspark.sql.functions import col, split

df.orderBy(col("salary").desc()) \
  .select(split(col("email"), "@").getItem(1).alias("email_domain")) \
  .show()
```

<b> What is the first name of the employee with the email address that contains the substring 'hr'?

```python
# Write your code using the Spark Function
from pyspark.sql.functions import col

df.filter(col("email").contains("hr")) \
  .select("first_name", "email") \
  .show()
```

d
<b> What is the last name of the employee with the email address that ends with the domain '.com'?

```python
# Write your code using the Spark Function
df.filter(col("email").endswith(".com")) \
  .select("last_name", "email") \
  .show()
```

d
<b> How many employees have an email address that contains their last name?

```python
# Write your code using the Spark Function
from pyspark.sql.functions import lower, col

df.filter(
    lower(col("email")).contains(lower(col("last_name")))
).count()
```

<b> What is the email of the employee with the first name that contains a space character?

```python
# Write your code using the Spark Function
from pyspark.sql.functions import col

df.filter(col("first_name").contains(" ")) \
  .select("first_name", "email") \
  .show()
```

<b> What is the email of the employee with the last name that contains a hyphen character?

```python
# Write your code using the Spark Function
from pyspark.sql.functions import col

df.filter(col("last_name").contains("-")) \
  .select("last_name", "email") \
  .show()
```
